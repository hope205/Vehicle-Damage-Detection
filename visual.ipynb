```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# your confusion matrix values (rows=predicted, cols=true)
matrix = np.array([
    [191, 57, 0, 3908],
    [35, 234, 2, 4923],
    [1, 0, 101, 83],
    [9, 16, 1, 0]
])

labels = ["dent", "scratch", "clean", "background"]

fig, ax = plt.subplots(figsize=(8, 6))
sns.heatmap(
    matrix,
    annot=True,
    fmt="d",
    cmap="viridis",       # better perceptual contrast than Blues
    xticklabels=labels,
    yticklabels=labels,
    cbar=True,
    linewidths=0.5,
    linecolor="white",
    annot_kws={"size": 10, "weight": "bold"}
)
ax.set_xlabel("True")
ax.set_ylabel("Predicted")
plt.title("Confusion Matrix")
plt.tight_layout()
plt.savefig("confusion_matrix.png", dpi=150)
```
